## 3.2 Export des tables

```python
from sqlalchemy import types

# 1. Catégorie des produits (aucune dépendance)
Cat_Prds.to_sql('Categorie_Produits', engine, if_exists='replace', index=False, dtype={
    'Category_id': types.VARCHAR(20),
    'Product_Category': types.NVARCHAR(100),
    'Category_Image': types.NVARCHAR(500)
})
print("Categorie_Produits exportée")
```

```python
# 2 . Fournisseurs : 
Frs.to_sql('Fournisseur' , engine , if_exists='replace' , index = False, dtype={
    'Fournisseur_id': types.VARCHAR(20),
    'Fournisseur_Nom': types.NVARCHAR(255),
    'Founisseur_image': types.NVARCHAR(500)
})
print('fournisseur exportée')
```

```python
# 3. Magasins (aucune dépendance)
Mag.to_sql('Magasins', engine, if_exists='replace', index=False, dtype={
    'magasin_id': types.VARCHAR(20),
    'Zone': types.NVARCHAR(255),
    'Ville': types.NVARCHAR(100),
    'Region': types.NVARCHAR(100),
    'Adresse': types.NVARCHAR(255)
})
print("Magasins exportée")
```

```python
# 4. Produits (dépend de Fournisseurs + Categorie_Produits)
Prds.to_sql('Produits', engine, if_exists='replace', index=False, dtype={
    'Produit_SKU': types.VARCHAR(20),
    'Produit_designation': types.NVARCHAR(255),
    'Prix_Achat': types.DECIMAL(10,2),
    'Prix_vente': types.DECIMAL(10,2),
    'Fournisseur_id': types.VARCHAR(20),
    'Category_id': types.VARCHAR(20)
})
print("Produits exportée")
```

```python
# 5. Transactions (dépend de Produits + Magasins)
Trans.to_sql('Transactions', engine, if_exists='replace', index=False, dtype={
    'transaction_id': types.INTEGER(),
    'transaction_id_original': types.INTEGER(),
    'Date': types.DATETIME(),
    'Produit_SKU': types.VARCHAR(20),
    'Quantite': types.INTEGER(),
    'Prix_Vente': types.DECIMAL(10,2),
    'Ventes': types.DECIMAL(10,2),
    'magasin_id': types.VARCHAR(20)
})
print("Transactions exportée")
```
